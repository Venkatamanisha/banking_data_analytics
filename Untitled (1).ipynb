```python
bins = [0,100000,300000,float('inf')]
labels = ['Low','Med','High']
df['Income_Band'] = pd.cut(df['estimated_income'],bins = bins, labels = labels, right = False)
```

```python


counts = df['Income_Band'].value_counts()
ax = counts.plot(kind='bar', color=['skyblue','orange','green'])

# Add labels on top of bars
for i, v in enumerate(counts):
    ax.text(i, v + 10, str(v), ha='center', fontweight='bold')

plt.xlabel("Income Band")
plt.ylabel("Count")
plt.title("Customer Distribution by Income Band")
plt.show()
```

# Numerical Analysis

```python
numerical_cols = ['estimated_income','credit_card_balance']
plt.figure(figsize = (12,10))
for i,col in enumerate(numerical_cols):
    plt.subplot(2,1,i+1)
    plt.title(col)
    sns.histplot(df[col], bins=20, kde=True, color='blue')
    
plt.show()
```

# Heat Maps

```python
numerical_cols = ['estimated_income','credit_card_balance','superannuation_savings','bank_loans',
                  'bank_deposits','checking_accounts','saving_accounts','foreign_currency_account','business_lending']
correlation_matrix = df[numerical_cols].corr()

plt.figure(figsize = (12,12))
sns.heatmap(correlation_matrix,annot = True,cmap = 'crest',fmt = '.2f')
plt.title("Correlation Matrix")
```
